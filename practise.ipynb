```python
import json
```

```python
with open('tickets-training-20.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
customer_texts = []
for ticket in data:
    ticket_id = ticket['conversationNo']
    customer_name = ticket['customerName']
    texts = []
    for msg in ticket.get('messages', []):
        if msg.get('senderType') == 'customer':
            texts.append(msg.get('text', ''))
    if texts:
        customer_texts.append({
            'conversationNo': ticket_id,
            'customerName': customer_name,
            'customer_texts': texts
        })
```

```python
print(f"Extracted {len(customer_texts)} tickets with customer texts.")
print("First ticket example:")
print(customer_texts[0] if customer_texts else "No data")
```

```python
for i in range(5):
    print(f"Customer Texts for Ticket {customer_texts[i]['conversationNo']}:")
    for text in customer_texts[i]['customer_texts']:
        print(f"- {text}")
```
